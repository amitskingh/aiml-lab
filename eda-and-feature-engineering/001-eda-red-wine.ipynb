```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 10))

sns.heatmap(
    df.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

# plt.xticks(rotation=45, ha="right")
# plt.yticks(rotation=0)
# plt.tight_layout()
plt.show()
```

```python
df.quality.value_counts()
# conclusion: the dataset is imbalanced, with more wines having a quality of 5 and 6 than any other quality score.
```

```python
df.quality.value_counts().plot(kind='bar', figsize=(10, 6), color='skyblue')
plt.xlabel('Wine Quality')
plt.ylabel('Count')
plt.title('Distribution of Wine Quality')
plt.show()
```

```python
# Distribution of each feature
df.head()
```

```python
for column in df.columns:
    sns.histplot(df[column], kde=True)
    plt.title(f'Distribution of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.show()
```

```python
# univariate analysis: analyzing each feature individually to understand its distribution and characteristics.
sns.pairplot(df, hue='quality', palette='Set1')
```

```python
sns.catplot(x='quality', y='alcohol', data=df, kind='box', height=6, aspect=1.5)
```

```python
sns.scatterplot(x='alcohol', y='density', hue='quality', data=df, palette='Set1')
```
